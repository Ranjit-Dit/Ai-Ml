```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
dataset = pd.read_csv('Salary Data.csv',usecols = ['Years of Experience','Salary'])
dataset = dataset.dropna()     #drops rows having empty value or NAN value
dataset.head()
```

```python
# dataset[['Years of Experience']]
```

```python
# dataset['Years of Experience'].values
```

```python
X = dataset.iloc[:,:1].values

# X = dataset['Years of Experience']
X
```

```python
Y = dataset.iloc[:,1:].values
# Y = dataset['Salary']
Y
```

# Splitting train and test

4_train-test-split.webp

```python
# train size-> min 2/3
# test size -> 1/3

from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test = train_test_split(X,Y,test_size = 0.1,random_state = 1)
```

```python
# print(Y_train)
```

```python
total_len = (len(X))
print("All total length = ",total_len)
print("X_train length = ", len(X_train))
print("X_test length = ", len(X_test))
print("Y_train length = ", len(Y_train))
print("Y_test length = ", len(Y_test))
```

```python
#Training model
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(X_train,Y_train)
```
